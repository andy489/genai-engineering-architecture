### Setup

```python
import time
import torch
import numpy as np
from transformers import AutoTokenizer, AutoModelForCausalLM

device = "cpu"

tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained("distilgpt2").to(device)
model.eval()

print("Device:", device)
```

### Helper Functions (Metrics + Inference)

```python
def run_inference(prompts, max_new_tokens=30):
    inputs = tokenizer(
        prompts,
        return_tensors="pt",
        padding=True
    ).to(device)

    start = time.time()
    with torch.no_grad():
        _ = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens
        )
    end = time.time()

    return end - start


def benchmark(batch_size, num_requests=20):
    prompts = ["Explain batching in AI."] * batch_size
    latencies = []

    for _ in range(num_requests):
        latency = run_inference(prompts)
        latencies.append(latency)

    latencies = np.array(latencies)

    return {
        "batch_size": batch_size,
        "avg_latency_sec": latencies.mean(),
        "p50_latency_sec": np.percentile(latencies, 50),
        "p95_latency_sec": np.percentile(latencies, 95),
        "throughput_req_per_sec": batch_size / latencies.mean()
    }
```

### Single vs Batched Requests

```python
results = []

for batch_size in [1, 2, 4, 8]:
    metrics = benchmark(batch_size)
    results.append(metrics)
    print(metrics)
```
